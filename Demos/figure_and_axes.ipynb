# Figure and Axes in Matplotlib

This notebook explores the fundamental building blocks of matplotlib: the Figure and Axes objects. Understanding the relationship between these two objects is crucial for mastering matplotlib's object-oriented interface.

## Key Concepts
- **Figure**: The top-level container for all plot elements
- **Axes**: The area where data is plotted (not to be confused with "axis")

```python
# Import statements
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
```

## Basic Figure Parameters

The figure object is the top-level container. Common parameters include:
- `figsize`: A tuple specifying (width, height) in inches
- `facecolor`: The background color of the figure
- `dpi`: Dots per inch (resolution)

```python
# Basic figure creation with parameters
fig = plt.figure(figsize=(8, 6), facecolor='lightgray', dpi=100)
ax = fig.add_subplot(111)
ax.text(0.5, 0.5, 'Figure with custom parameters', 
        ha='center', va='center', fontsize=14)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
plt.show()
```

## Figure vs Axes Facecolor

This example from the book illustrates the difference between figure and axes by setting different background colors:

```python
# Figure vs Axes facecolor demonstration
fig = plt.figure(figsize=(2, 3), facecolor='gray')
ax = plt.axes(facecolor='lightyellow')
plt.show()
```

Notice how the gray figure background extends beyond the yellow axes area. The axes is where your data gets plotted, while the figure includes everything (axes, labels, title, etc.).

## Figure Methods

While most plotting is done through axes methods, the figure object has several useful methods:

```python
# Figure methods demonstration
fig = plt.figure(figsize=(10, 6))

# Add multiple subplots
ax1 = fig.add_subplot(2, 2, 1)
ax2 = fig.add_subplot(2, 2, 2)
ax3 = fig.add_subplot(2, 2, 3)
ax4 = fig.add_subplot(2, 2, 4)

# Add a main title for the entire figure
fig.suptitle('Figure with Multiple Subplots', fontsize=16)

# Plot different things in each subplot
x = np.linspace(0, 2*np.pi, 100)
ax1.plot(x, np.sin(x))
ax1.set_title('Sine Wave')

ax2.plot(x, np.cos(x))
ax2.set_title('Cosine Wave')

ax3.plot(x, np.sin(x) * np.cos(x))
ax3.set_title('Sin × Cos')

ax4.plot(x, np.sin(x)**2)
ax4.set_title('Sin²')

# Adjust spacing between subplots
fig.tight_layout()
plt.show()
```

## Axes Methods and Getters/Setters

The axes object is where most of the action happens. Many axes methods come in getter/setter pairs:

```python
# Demonstrating getters and setters
x = np.linspace(0, 1, 2)
fig, ax = plt.subplots(figsize=(8, 4))

# Plot some data
ax.plot(x, x, label='y = x')
ax.plot(x, 1 - x, label='y = 1 - x')

# Set title (setter)
ax.set_title("My Chart")
print(f"Title via attribute: {ax.title}")
print(f"Title via getter: {ax.get_title()}")

# Update title
ax.set_title("My Wholesome Chart")
print(f"Updated title: {ax.get_title()}")

# Other getter/setter examples
ax.set_xlim([-0.1, 1.1])
print(f"X limits: {ax.get_xlim()}")

ax.set_ylabel("Y values")
print(f"Y label: {ax.get_ylabel()}")

ax.legend()
plt.show()
```

## Common Axes Methods

Here's a comprehensive example showing various axes methods:

```python
# Create figure and axes
fig, ax = plt.subplots(figsize=(10, 6))

# Generate some data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Plotting methods
ax.plot(x, y1, 'b-', label='sin(x)', linewidth=2)
ax.plot(x, y2, 'r--', label='cos(x)', linewidth=2)

# Axes configuration methods
ax.set_title('Trigonometric Functions', fontsize=16, fontweight='bold')
ax.set_xlabel('x values', fontsize=12)
ax.set_ylabel('y values', fontsize=12)

# Set limits
ax.set_xlim([0, 10])
ax.set_ylim([-1.5, 1.5])

# Grid
ax.grid(True, alpha=0.3)

# Add horizontal and vertical lines
ax.axhline(y=0, color='k', linestyle='-', alpha=0.3)
ax.axvline(x=5, color='g', linestyle=':', alpha=0.5)

# Add text annotation
ax.text(5, 1.2, 'x = 5', ha='center', fontsize=10)

# Legend
ax.legend(loc='upper right')

# Customize ticks
ax.set_xticks(np.arange(0, 11, 2))
ax.set_yticks([-1, -0.5, 0, 0.5, 1])

plt.show()
```

## Interactive Exercise 3: Figure and Axes Together

Create a figure with multiple subplots showing different mathematical functions:

```python
# TODO: Create a 2x2 grid of subplots showing:
# Top left: y = x
# Top right: y = x²
# Bottom left: y = x³
# Bottom right: y = √x

fig = plt.figure(figsize=(10, 8))

# Generate x values
x = np.linspace(0, 10, 100)

# TODO: Create subplot 1 (top left) - Linear
ax1 = fig.add_subplot(2, 2, 1)
ax1.plot(x, x)
ax1.set_title('Linear: y = x')
ax1.grid(True, alpha=0.3)

# TODO: Create subplot 2 (top right) - Quadratic
ax2 = fig.add_subplot(2, 2, 2)
ax2.plot(x, x**2)
ax2.set_title('Quadratic: y = x²')
ax2.grid(True, alpha=0.3)

# TODO: Create subplot 3 (bottom left) - Cubic
ax3 = fig.add_subplot(2, 2, 3)
ax3.plot(x, x**3)
ax3.set_title('Cubic: y = x³')
ax3.grid(True, alpha=0.3)

# TODO: Create subplot 4 (bottom right) - Square root
ax4 = fig.add_subplot(2, 2, 4)
ax4.plot(x, np.sqrt(x))
ax4.set_title('Square Root: y = √x')
ax4.grid(True, alpha=0.3)

# Add a main title and adjust layout
fig.suptitle('Common Mathematical Functions', fontsize=16)
fig.tight_layout()

plt.show()
```

## Summary

Key takeaways:
1. **Figure** is the top-level container; **Axes** is where data is plotted
2. Most plotting is done through axes methods (ax.plot(), ax.scatter(), etc.)
3. Many axes methods come in getter/setter pairs (get_title/set_title)
4. Figure methods are useful for managing multiple subplots and overall appearance
5. Understanding the figure/axes relationship is crucial for the object-oriented interface
