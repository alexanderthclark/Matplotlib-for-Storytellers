# Mixing MATLAB-style and Object-Oriented Interfaces

While matplotlib's object-oriented interface is more powerful, you can mix both interfaces when convenient. This notebook explores how to effectively combine pyplot (MATLAB-style) functions with the OO approach.

## Key Functions for Mixing
- `plt.gca()`: Get Current Axes
- `plt.gcf()`: Get Current Figure
- `plt.sca(ax)`: Set Current Axes

```python
# Import statements
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## Starting with MATLAB-style, Converting to OO

This example from the book shows how to start with pyplot and then switch to OO:

```python
# Start with MATLAB-style
x = np.linspace(0, 1, 2)
plt.plot(x, x)
plt.title("My Chart")

# Get the current axes object
ax = plt.gca()
print(f"Title from axes: {ax.title}")

# Now use OO interface
ax.plot(x, 1 - x)
ax.set_title('My Wholesome Chart')
print(f"Updated title: {ax.title}")

plt.show()
```

## Starting with OO, Using pyplot Functions

The colorful example from the book demonstrates starting with OO and mixing in pyplot:

```python
# OOP Start
fig, ax = plt.subplots(figsize=(8, 5))

x = np.linspace(0, 100, 2)
ax.plot(x, x, color='gray')

ax.set_xlim([0, 100])
ax.set_ylim([0, 100])

# Back to pyplot functions
for i in range(0, 101, 5):  # Reduced for performance
    plt.axvline(i, 0, i/100, color=f'C{i%10}')
    plt.axvline(i, i/100, 1, color=f'C{(i+5)%10}')

plt.axis('off')
plt.show()
```

## Understanding Current Axes

When you use pyplot functions, they operate on the "current" axes:

```python
# Create multiple subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Plot on first axes using OO
ax1.plot([1, 2, 3], [1, 4, 9])
ax1.set_title('First Axes (OO)')

# Make ax2 the current axes
plt.sca(ax2)

# Now pyplot functions will operate on ax2
plt.plot([1, 2, 3], [1, 2, 3])
plt.title('Second Axes (pyplot)')
plt.xlabel('X axis')

# Verify current axes
current = plt.gca()
print(f"Current axes is ax2: {current is ax2}")
print(f"Current axes is ax1: {current is ax1}")

plt.tight_layout()
plt.show()
```

## Common Mixed Patterns

Here are some common scenarios where mixing interfaces is useful:

### Pattern 1: Quick Styling with pyplot

```python
# Create plot with OO
fig, ax = plt.subplots(figsize=(8, 6))

# Plot data
x = np.linspace(0, 10, 100)
ax.plot(x, np.sin(x), label='sin(x)')
ax.plot(x, np.cos(x), label='cos(x)')

# Use pyplot for quick styling
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()

plt.show()
```

### Pattern 2: Saving Figures

```python
# These three approaches are equivalent for saving
fig, ax = plt.subplots()
ax.plot([1, 2, 3], [1, 4, 9])

# Method 1: Using figure object
# fig.savefig('plot1.png')

# Method 2: Using pyplot
# plt.savefig('plot2.png')

# Method 3: Getting current figure
# current_fig = plt.gcf()
# current_fig.savefig('plot3.png')

print("All three methods save the same figure!")
plt.close()
```

### Pattern 3: Working with Colorbars

```python
# Colorbars often require mixing interfaces
fig, ax = plt.subplots(figsize=(8, 6))

# Create a heatmap
data = np.random.randn(10, 10)
im = ax.imshow(data, cmap='coolwarm')
ax.set_title('Random Heatmap')

# Add colorbar using pyplot (common pattern)
plt.colorbar(im, ax=ax)

plt.show()
```

## Pitfalls When Mixing Interfaces

Be careful when mixing interfaces, especially with subplots:

```python
# Potential pitfall: pyplot functions affect the last created axes
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Plot on both axes
ax1.plot([1, 2, 3], [1, 2, 3])
ax2.plot([1, 2, 3], [3, 2, 1])

# This affects ax2 (the last created axes)!
plt.title('This title goes to ax2!')
plt.ylabel('This label also goes to ax2!')

# To affect ax1, we need to make it current first
plt.sca(ax1)
plt.title('Now this goes to ax1')

plt.tight_layout()
plt.show()
```

## Best Practices for Mixing

1. **Be explicit about which axes you're working with**
2. **Prefer OO methods when working with multiple subplots**
3. **Use pyplot for figure-level operations (savefig, show, tight_layout)**
4. **Document when and why you're switching interfaces**

```python
# Example following best practices
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()  # Make it easier to iterate

# Use OO for subplot-specific operations
for i, ax in enumerate(axes):
    x = np.linspace(0, 2*np.pi, 100)
    ax.plot(x, np.sin(x + i*np.pi/2))
    ax.set_title(f'Phase shift: {i}π/2')
    ax.grid(True, alpha=0.3)

# Use pyplot for figure-level operations
plt.suptitle('Sine Waves with Different Phase Shifts', fontsize=16)
plt.tight_layout()
plt.show()
```

## Interactive Exercise 2: Strategic Mixing

Create a complex visualization using both interfaces strategically:

```python
# TODO: Create a 2x2 subplot figure that:
# 1. Uses OO interface for subplot-specific operations
# 2. Uses pyplot for figure-level operations
# 3. Demonstrates switching between current axes

# Create the figure
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 8))

# Generate data
x = np.linspace(0, 10, 100)

# Subplot 1: Use pure OO
ax1.plot(x, np.sin(x))
ax1.set_title('Pure OO: Sine')
ax1.grid(True, alpha=0.3)

# Subplot 2: Start with OO, add pyplot styling
ax2.plot(x, np.cos(x))
plt.sca(ax2)  # Make ax2 current
plt.title('Mixed: Cosine')
plt.grid(True, alpha=0.3)

# Subplot 3: Use pyplot after setting current axes
plt.sca(ax3)
plt.plot(x, np.exp(-x/5) * np.sin(x))
plt.title('Pyplot: Damped Sine')
plt.grid(True, alpha=0.3)

# Subplot 4: Mix creatively
ax4.scatter(x[::5], np.random.randn(20), alpha=0.6)
plt.sca(ax4)
plt.axhline(y=0, color='k', linestyle='--', alpha=0.5)
ax4.set_title('Creative Mix: Random Scatter')

# Figure-level operations with pyplot
plt.suptitle('Demonstrating Mixed Interface Usage', fontsize=16)
plt.tight_layout()
plt.show()
```

## Advanced: Context Managers

For temporary style changes, you can use context managers:

```python
# Using context managers for temporary style changes
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0, 10, 100)
y = np.sin(x)

# Normal plot
ax1.plot(x, y)
ax1.set_title('Default Style')

# Plot with temporary style
with plt.style.context('seaborn-v0_8-darkgrid'):
    ax2.plot(x, y)
    ax2.set_title('With Seaborn Style Context')

plt.tight_layout()
plt.show()
```

## Summary

Key takeaways about mixing interfaces:

1. **plt.gca()** and **plt.gcf()** bridge between interfaces
2. **plt.sca(ax)** sets the current axes for pyplot functions
3. **pyplot functions** operate on the current axes/figure
4. **Be careful** with subplots - pyplot affects the current axes
5. **Best practice**: Use OO for axes operations, pyplot for figure operations
6. **Mixing is okay** when done thoughtfully and documented clearly

The ability to mix interfaces gives you flexibility, but with great power comes great responsibility!
