# Pandas Integration with Matplotlib

Pandas has excellent integration with matplotlib, providing convenient plotting methods directly on Series and DataFrame objects. This notebook explores how to leverage this integration while maintaining control through matplotlib's object-oriented interface.

## Key Concept
Pandas plotting methods can accept an `ax` parameter to specify which axes to plot on, allowing seamless integration with matplotlib's OO interface.

```python
# Import statements
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
import seaborn as sns  # For sample datasets
```

## Basic Pandas Plotting

Pandas provides a `plot()` method on both Series and DataFrame objects:

```python
# Create sample data
dates = pd.date_range('2023-01-01', periods=100)
df = pd.DataFrame({
    'A': np.cumsum(np.random.randn(100)),
    'B': np.cumsum(np.random.randn(100)),
    'C': np.cumsum(np.random.randn(100))
}, index=dates)

# Basic pandas plot
df.plot(figsize=(10, 6), title='Cumulative Random Walks')
plt.ylabel('Value')
plt.show()
```

## The Iris Box Plot Example from the Book

This demonstrates mixing pandas plotting with matplotlib's OO interface:

```python
# Load iris dataset
from sklearn.datasets import load_iris
data = load_iris()
df_iris = pd.DataFrame(data['data'], columns=data['feature_names'])

# Create figure and axes first
fig, ax = plt.subplots(figsize=(8, 6))

# Use pandas plot method with ax parameter
df_iris.plot.box(ax=ax)

# Customize with axes methods
ax.set_title('Iris Dataset Features', fontsize=14)
ax.yaxis.grid(True)
ax.xaxis.grid(False)
ax.set_ylabel('Measurement (cm)')

# Use pyplot for figure-level adjustments
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## Different Plot Types with Pandas

Pandas supports many plot types, all of which can be integrated with matplotlib:

```python
# Create a figure with subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

# Sample data
df_sample = pd.DataFrame({
    'A': np.random.randn(50),
    'B': np.random.randn(50) + 1,
    'C': np.random.randn(50) + 2,
    'category': np.random.choice(['X', 'Y', 'Z'], 50)
})

# 1. Line plot
df_sample[['A', 'B', 'C']].plot(ax=axes[0])
axes[0].set_title('Line Plot')
axes[0].set_xlabel('Index')

# 2. Bar plot
df_sample[['A', 'B', 'C']].iloc[:10].plot.bar(ax=axes[1])
axes[1].set_title('Bar Plot')
axes[1].set_xlabel('Index')

# 3. Histogram
df_sample[['A', 'B', 'C']].plot.hist(ax=axes[2], alpha=0.7, bins=15)
axes[2].set_title('Histogram')
axes[2].set_xlabel('Value')

# 4. Scatter plot
df_sample.plot.scatter(x='A', y='B', c='C', colormap='viridis', ax=axes[3])
axes[3].set_title('Scatter Plot')

plt.tight_layout()
plt.show()
```

## Series Plotting

Series objects also have plotting methods:

```python
# Create a time series
ts = pd.Series(
    np.cumsum(np.random.randn(365)),
    index=pd.date_range('2023-01-01', periods=365)
)

# Create custom figure
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# Plot the series
ts.plot(ax=ax1, color='blue', alpha=0.7)
ax1.set_title('Daily Values')
ax1.set_ylabel('Value')
ax1.grid(True, alpha=0.3)

# Plot rolling mean
ts.rolling(window=30).mean().plot(ax=ax2, color='red', label='30-day MA')
ts.rolling(window=7).mean().plot(ax=ax2, color='green', label='7-day MA')
ax2.set_title('Moving Averages')
ax2.set_ylabel('Value')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## GroupBy Plotting

Pandas groupby operations integrate well with plotting:

```python
# Create sample sales data
np.random.seed(42)
dates = pd.date_range('2023-01-01', periods=365)
sales_data = pd.DataFrame({
    'date': dates,
    'product': np.random.choice(['A', 'B', 'C'], 365),
    'region': np.random.choice(['North', 'South', 'East', 'West'], 365),
    'sales': np.random.randint(100, 1000, 365)
})

# Group by product and plot
fig, ax = plt.subplots(figsize=(10, 6))

sales_data.groupby('product')['sales'].sum().plot.bar(ax=ax)
ax.set_title('Total Sales by Product', fontsize=14)
ax.set_ylabel('Total Sales')
ax.set_xlabel('Product')

# Add value labels on bars
for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

## Advanced Integration: Custom Styling

```python
# Create more complex data
df_complex = pd.DataFrame({
    'Quarter': ['Q1', 'Q2', 'Q3', 'Q4'] * 3,
    'Year': [2021, 2021, 2021, 2021, 2022, 2022, 2022, 2022, 2023, 2023, 2023, 2023],
    'Revenue': [100, 120, 140, 160, 110, 135, 155, 180, 125, 150, 175, 200],
    'Profit': [20, 25, 30, 35, 22, 28, 33, 40, 26, 32, 38, 45]
})

# Pivot for easier plotting
revenue_pivot = df_complex.pivot(index='Quarter', columns='Year', values='Revenue')

# Create custom styled plot
fig, ax = plt.subplots(figsize=(10, 6))

# Use pandas plot with custom styling
revenue_pivot.plot(kind='bar', ax=ax, color=['#FF6B6B', '#4ECDC4', '#45B7D1'], width=0.8)

# Customize with matplotlib
ax.set_title('Quarterly Revenue by Year', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Quarter', fontsize=12)
ax.set_ylabel('Revenue ($M)', fontsize=12)
ax.legend(title='Year', frameon=False, loc='upper left')
ax.grid(True, axis='y', alpha=0.3)
ax.set_axisbelow(True)

# Customize spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Rotate x labels
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Interactive Exercise 1: Combining Multiple Pandas Plots

Create a dashboard-style visualization combining different pandas plot types:

```python
# TODO: Create a 2x2 dashboard showing:
# 1. Time series line plot
# 2. Category distribution (bar plot)
# 3. Correlation heatmap
# 4. Distribution histogram

# Generate sample data
np.random.seed(42)
n_days = 90
dates = pd.date_range('2023-01-01', periods=n_days)

df_dashboard = pd.DataFrame({
    'date': dates,
    'sales': np.cumsum(np.random.randn(n_days)) + 100,
    'customers': np.random.poisson(50, n_days),
    'category': np.random.choice(['Electronics', 'Clothing', 'Food'], n_days),
    'satisfaction': np.random.uniform(3, 5, n_days)
})

# Create the dashboard
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(14, 10))

# 1. Time series
df_dashboard.set_index('date')['sales'].plot(ax=ax1, color='navy')
ax1.set_title('Sales Over Time')
ax1.set_ylabel('Sales')
ax1.grid(True, alpha=0.3)

# 2. Category distribution
df_dashboard.groupby('category')['sales'].count().plot.bar(ax=ax2, color='skyblue')
ax2.set_title('Sales by Category')
ax2.set_ylabel('Count')
ax2.set_xlabel('Category')

# 3. Correlation heatmap (using scatter as proxy)
df_dashboard.plot.scatter(x='customers', y='satisfaction', c='sales', 
                         colormap='viridis', ax=ax3)
ax3.set_title('Customer Satisfaction vs Count')

# 4. Distribution
df_dashboard['satisfaction'].plot.hist(bins=20, ax=ax4, color='green', alpha=0.7)
ax4.set_title('Satisfaction Score Distribution')
ax4.set_xlabel('Satisfaction Score')
ax4.set_ylabel('Frequency')

plt.suptitle('Sales Dashboard', fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```

## Interactive Exercise 2: Custom Pandas Plot Function

Create a reusable function that combines pandas and matplotlib:

```python
# TODO: Create a function that:
# 1. Takes a DataFrame and column names
# 2. Creates a figure with pandas plots
# 3. Adds custom matplotlib styling
# 4. Returns the figure and axes

def create_comparison_plot(df, col1, col2, title="Comparison Plot"):
    """
    Create a comparison plot with two columns from a DataFrame.
    
    Parameters:
    df: pandas DataFrame
    col1, col2: column names to compare
    title: plot title
    
    Returns:
    fig, axes: matplotlib figure and axes
    """
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))
    
    # Plot 1: Line comparison
    df[[col1, col2]].plot(ax=ax1)
    ax1.set_title(f'{col1} vs {col2} Over Time')
    ax1.legend()
    
    # Plot 2: Scatter plot
    df.plot.scatter(x=col1, y=col2, ax=ax2, alpha=0.6)
    ax2.set_title(f'{col1} vs {col2} Scatter')
    
    # Plot 3: Distribution comparison
    df[[col1, col2]].plot.hist(ax=ax3, alpha=0.7, bins=20)
    ax3.set_title('Distribution Comparison')
    ax3.legend()
    
    # Overall styling
    fig.suptitle(title, fontsize=16)
    
    for ax in [ax1, ax2, ax3]:
        ax.grid(True, alpha=0.3)
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
    
    plt.tight_layout()
    return fig, (ax1, ax2, ax3)

# Test the function
test_df = pd.DataFrame({
    'A': np.cumsum(np.random.randn(100)),
    'B': np.cumsum(np.random.randn(100)) * 1.5 + 10
})

fig, axes = create_comparison_plot(test_df, 'A', 'B', "Custom Comparison Plot")
plt.show()
```

## Best Practices for Pandas-Matplotlib Integration

1. **Always use the `ax` parameter** when you need control over the plot
2. **Create figure and axes first** when making subplots
3. **Use pandas for convenience**, matplotlib for customization
4. **Remember that pandas returns axes objects** which you can further customize
5. **Check the pandas visualization documentation** for plot-specific parameters

```python
# Example of best practices
# Create sample financial data
dates = pd.date_range('2020-01-01', periods=1000)
stock_data = pd.DataFrame({
    'AAPL': 100 + np.cumsum(np.random.randn(1000)) * 2,
    'GOOGL': 1000 + np.cumsum(np.random.randn(1000)) * 10,
    'MSFT': 150 + np.cumsum(np.random.randn(1000)) * 3
}, index=dates)

# Normalize for comparison
normalized = stock_data / stock_data.iloc[0] * 100

# Create professional-looking plot
fig, ax = plt.subplots(figsize=(12, 6))

# Use pandas for plotting
normalized.plot(ax=ax, linewidth=2)

# Customize with matplotlib
ax.set_title('Normalized Stock Performance (Base = 100)', fontsize=16, pad=20)
ax.set_xlabel('Date', fontsize=12)
ax.set_ylabel('Normalized Price', fontsize=12)
ax.legend(title='Stock', frameon=False, loc='upper left')
ax.grid(True, alpha=0.3)
ax.set_facecolor('#f8f9fa')

# Add reference line
ax.axhline(y=100, color='black', linestyle='--', alpha=0.5, linewidth=1)
ax.text(dates[10], 101, 'Base (100)', fontsize=10)

# Format y-axis as percentage
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y:.0f}'))

plt.tight_layout()
plt.show()
```

## Summary

Key takeaways about pandas-matplotlib integration:

1. **Pandas plot() methods** are convenient wrappers around matplotlib
2. **The ax parameter** is your bridge between pandas and matplotlib OO interface
3. **Different plot types** (line, bar, scatter, hist, box) all support the ax parameter
4. **GroupBy operations** integrate seamlessly with plotting
5. **Best practice**: Let pandas handle the data transformation, matplotlib handle the styling
6. **Return values**: Pandas plot methods return axes objects for further customization

This integration allows you to leverage the best of both worlds: pandas' data handling and matplotlib's customization!
