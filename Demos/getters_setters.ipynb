# Getters and Setters in Matplotlib

This notebook explores the getter and setter pattern in matplotlib's object-oriented interface. Understanding this pattern is key to effectively manipulating plot properties.

## Why Getters and Setters?

Matplotlib uses getter and setter methods to access and modify object properties. While this might seem old-fashioned compared to direct attribute access, it provides consistency and control over how properties are managed.

```python
# Import statements
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## Basic Getter/Setter Pattern

Most axes properties follow a consistent pattern:
- `set_property()` to set a value
- `get_property()` to retrieve the value

```python
# Create a simple plot
fig, ax = plt.subplots()

# Setting properties
ax.set_title("Original Title")
ax.set_xlabel("X Label")
ax.set_ylabel("Y Label")

# Getting properties
print(f"Title: {ax.get_title()}")
print(f"X Label: {ax.get_xlabel()}")
print(f"Y Label: {ax.get_ylabel()}")

plt.close()  # Close to avoid displaying empty plot
```

## Title: A Special Case

The title property is unique - it can also be accessed as an attribute:

```python
fig, ax = plt.subplots()

# Set title
ax.set_title("My Chart")

# Three ways to access the title
print(f"Via attribute: {ax.title}")
print(f"Via getter: {ax.get_title()}")
print(f"Direct from Text object: {ax.title.get_text()}")

# Note: ax.title is actually a Text object
print(f"\nType of ax.title: {type(ax.title)}")

plt.close()
```

## Setters Overwrite, Plot Methods Accumulate

This is a crucial distinction from the book example:

```python
# Demonstrating the difference between setters and plot methods
x = np.linspace(0, 1, 2)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Left plot: Multiple plot() calls accumulate
ax1.plot(x, x, 'b-', label='First plot')
ax1.plot(x, 1 - x, 'r-', label='Second plot')
ax1.set_title("plot() calls accumulate")
ax1.legend()

# Right plot: Multiple set_title() calls overwrite
ax2.plot(x, x, 'g-')
ax2.set_title("First Title")
print(f"After first set_title: '{ax2.get_title()}'")

ax2.set_title("Second Title")
print(f"After second set_title: '{ax2.get_title()}'")

ax2.set_title("Final Title")
print(f"Final title: '{ax2.get_title()}'")

plt.tight_layout()
plt.show()
```

## Common Getter/Setter Pairs

Here's a comprehensive list of commonly used getter/setter pairs:

```python
fig, ax = plt.subplots(figsize=(8, 6))

# Plot some data first
x = np.linspace(0, 10, 100)
y = np.sin(x)
ax.plot(x, y)

# Demonstrate various getter/setter pairs
properties = [
    ('title', 'Sine Wave'),
    ('xlabel', 'Time (s)'),
    ('ylabel', 'Amplitude'),
    ('xlim', (0, 10)),
    ('ylim', (-1.5, 1.5)),
    ('xticks', [0, 2, 4, 6, 8, 10]),
    ('xticklabels', ['0', '2π/5', '4π/5', '6π/5', '8π/5', '2π'])
]

print("Setting and getting properties:\n")
for prop, value in properties:
    # Set the property
    setter = getattr(ax, f'set_{prop}')
    setter(value)
    
    # Get the property
    getter = getattr(ax, f'get_{prop}')
    retrieved = getter()
    
    print(f"{prop}:")
    print(f"  Set to: {value}")
    print(f"  Got back: {retrieved}\n")

plt.show()
```

## Properties Without Direct Attributes

Unlike `title`, most properties don't have direct attribute access:

```python
fig, ax = plt.subplots()

# Set some properties
ax.set_xlim([0, 10])
ax.set_ylim([-1, 1])

# This works:
print(f"X limits via getter: {ax.get_xlim()}")
print(f"Y limits via getter: {ax.get_ylim()}")

# But these don't exist as simple attributes:
try:
    print(f"X limits via attribute: {ax.xlim}")
except AttributeError as e:
    print(f"AttributeError: {e}")

# However, we can access the underlying objects:
print(f"\nX axis object: {ax.xaxis}")
print(f"Y axis object: {ax.yaxis}")

plt.close()
```

## Batch Setting with set()

You can set multiple properties at once using the `set()` method:

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

x = np.linspace(0, 2*np.pi, 100)

# Left: Setting properties one by one
line1, = ax1.plot(x, np.sin(x))
ax1.set_title('One by One')
ax1.set_xlabel('x')
ax1.set_ylabel('sin(x)')
ax1.set_xlim([0, 2*np.pi])
ax1.set_ylim([-1.2, 1.2])

# Right: Batch setting with set()
line2, = ax2.plot(x, np.cos(x))
ax2.set(
    title='Batch Setting',
    xlabel='x',
    ylabel='cos(x)',
    xlim=[0, 2*np.pi],
    ylim=[-1.2, 1.2]
)

# You can also use set() on line objects
line2.set(
    color='red',
    linewidth=2,
    linestyle='--',
    alpha=0.7
)

plt.tight_layout()
plt.show()
```

## Advanced: Getting All Artists

Sometimes you need to get all the objects (artists) that have been added to an axes:

```python
fig, ax = plt.subplots(figsize=(8, 6))

# Add various elements
x = np.linspace(0, 10, 100)
line1, = ax.plot(x, np.sin(x), label='sin')
line2, = ax.plot(x, np.cos(x), label='cos')
ax.axhline(y=0, color='k', linestyle='--', alpha=0.5)
ax.text(5, 0.5, 'Some text', fontsize=12)
ax.legend()

# Get all line objects
lines = ax.get_lines()
print(f"Number of lines: {len(lines)}")
for i, line in enumerate(lines):
    print(f"  Line {i}: {line.get_label()}")

# Get all text objects
texts = ax.texts
print(f"\nNumber of text objects: {len(texts)}")
for i, text in enumerate(texts):
    print(f"  Text {i}: '{text.get_text()}'")

# Get all children (everything)
print(f"\nTotal number of child artists: {len(ax.get_children())}")

plt.show()
```

## Summary

Key takeaways about getters and setters:

1. **Consistency**: Most properties follow the `set_property()`/`get_property()` pattern
2. **Setters overwrite**: Calling a setter multiple times replaces the previous value
3. **Plot methods accumulate**: Methods like `plot()` add to the axes without removing previous plots
4. **Not all properties have direct attributes**: You usually need to use getters, not attribute access
5. **Batch setting**: Use `set()` to update multiple properties at once
6. **Dynamic updates**: Setters enable interactive and animated visualizations

Understanding this pattern is crucial for effectively using matplotlib's object-oriented interface!
